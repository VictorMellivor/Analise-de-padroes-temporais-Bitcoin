```python
import pandas as pd
```

```python
dyahoo = pd.read_excel('../dataset/BTC Price History & Historical Data - Yahoo Finance.ods', engine='odf')
dcmc = pd.read_csv('../dataset/Bitcoin_01_01_2009-01_01_2025_historical_data_coinmarketcap.csv', delimiter=';')
dcoingecko = pd.read_csv('../dataset/btc-usd-max (coingecko).csv')
```

## 1. Formatação do dataset Yahoo

```python
dyahoo.dtypes
```

```python
dyahoo['Date'] = pd.to_datetime(dyahoo['Date'])
```

```python
dyahoo
```

## 2. Formatação do dataset CoinMarketCap

```python
dcmc_formatada = dcmc.drop(columns=['timeClose', 'timeHigh', 'timeLow', 'name', 'timestamp'])
```

```python
dcmc_formatada[['volume', 'marketCap']] = dcmc_formatada[['volume', 'marketCap']].astype(int)
```

```python
dcmc_formatada['timeOpen']
```

```python
dcmc_formatada['timeOpen'] = dcmc_formatada['timeOpen'].str.replace('T00:00:00.000Z', '')
```

```python
dcmc_formatada['timeOpen'] = pd.to_datetime(dcmc_formatada['timeOpen'])
dcmc_formatada['timeOpen'] = pd.to_datetime(dcmc_formatada['timeOpen'], format ='%d.%m.%Y')
```

```python
dcmc_formatada
```

## 3. Formatação do dataset Coingecko

```python
dcoingecko
```

```python
dcoingecko_formatado = dcoingecko.rename(columns={'snapped_at':'Data', 'market_cap':'Marketcap', 'total_volume':'Volume'})
```

```python
dcoingecko_formatado[['Marketcap', 'Volume']] = dcoingecko_formatado[['Marketcap', 'Volume']].astype(int)
```

```python
dcoingecko_formatado['Horário'] = dcoingecko_formatado['Data'].str[11:]
dcoingecko_formatado['Data'] = dcoingecko_formatado['Data'].str[:10]
```

```python
dcoingecko_formatado['Data'] = pd.to_datetime(dcoingecko_formatado['Data'])
dcoingecko_formatado['Data'] = pd.to_datetime(dcoingecko_formatado['Data'], format = '%Y.%m.%d')
```
