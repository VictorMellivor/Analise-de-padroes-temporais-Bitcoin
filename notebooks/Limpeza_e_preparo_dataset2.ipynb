## 2. Formatação do dataset CoinMarketCap

## Transformação de dados

### Drop, Rename & Replace

```python
dcmc_formatada = dcmc.drop(columns=['timeClose', 'timeHigh', 'timeLow', 'name', 'timestamp'])
```

```python
dcmc_formatada = dcmc_formatada.rename(columns={'timeOpen':'data'})
```

```python
dcmc_formatada['data'] = dcmc_formatada['data'].str.replace('T00:00:00.000Z', '')
```

### Criação de colunas

```python
dcmc_formatada['data'] = pd.to_datetime(dcmc_formatada['data'])
```

```python
dcmc_formatada['mês'] = dcmc_formatada['data'].dt.month
```

```python
dcmc_formatada['ano'] = dcmc_formatada['data'].dt.year
```

```python
calculo_variacao = dcmc_formatada['close']/dcmc_formatada['open']-1
dcmc_formatada['var%'] = calculo_variacao*100

dcmc_formatada['var%'] = dcmc_formatada['var%'].astype(str)
dcmc_formatada
```

### Alteração de tipos

```python
dcmc_formatada[['volume', 'marketCap']] = dcmc_formatada[['volume', 'marketCap']].astype(int)
```

```python
dcmc_formatada['volume'] = dcmc_formatada['volume'].astype(float)
dcmc_formatada['marketCap'] = dcmc_formatada['marketCap'].astype(float)
```

```python
dcmc_formatada.dtypes
```

```python
dcmc_formatada.sort_values(by= 'data', ascending=False)
```

### Funções para filtragem de meses negativos e positivos

```python
meses = list(range(1, 13))
```

```python
def EncontrarMes(df, valor):
    return df.loc[df['mês'].isin(valor)]
dados_brutos = EncontrarMes(dcmc_formatada, meses)
```

### Filtragem de valores negativos e positivos

```python
dataframe_negativo = pd.DataFrame(dados_brutos[dados_brutos['var%'].str.contains('-')])
dataframe_negativo['var%'] = dataframe_negativo['var%'].astype(float)
```

```python
dataframe_negativo[dataframe_negativo['ano'] == 2015]
```

```python
dataframe_con = pd.DataFrame(dados_brutos)
dataframe_con['var%'] = dataframe_con['var%'].astype(float)

dataframe_positivo = dataframe_con[dataframe_con['var%'] >= 0]
```

```python
dataframe_positivo[dataframe_positivo['ano'] == 2015]
```

## Média de volume anual

### Média anual de volume de meses com baixo desempenho
Os válores são influenciados pela quantidade de meses negativos e positivos de um determinado ano. Caso um ano tenha mais meses positivos do que negativos, a tendencia é que a **média de volume anual de meses com baixo desempenho** seja menor

```python
dataframe_negativo.groupby('ano')['volume'].mean()
```

### Média anual de volume de meses com bom desempenho 
Os válores são influenciados pela quantidade de meses negativos e positivos de um determinado ano. Caso um ano tenha mais meses negativos do que positivos, a tendencia é que a **média de volume anual de meses com bom desempenho** seja menor

```python
dataframe_positivo.groupby('ano')['volume'].mean()
```

### Média anual de variações

```python
dcmc_formatada['var%'] = dcmc_formatada['var%'].astype(float)
```

```python
dcmc_formatada.groupby('ano')['var%'].mean()
```

### Volume anual total 

```python
dataframe_negativo.groupby('ano')['volume'].sum()
```

```python
dataframe_positivo.groupby('ano')['volume'].sum()
```

```python
desenvolvimento_volume = px.line(dcmc_formatada, x='data', y='volume', log_y=True)
desenvolvimento_volume
```

## 3. Formatação do dataset Coingecko

```python
dcoingecko
```

```python
dcoingecko_formatado = dcoingecko.rename(columns={'snapped_at':'Data', 'market_cap':'Marketcap', 'total_volume':'Volume'})
```

```python
dcoingecko_formatado[['Marketcap', 'Volume']] = dcoingecko_formatado[['Marketcap', 'Volume']].astype(int)
```

```python
dcoingecko_formatado['Horário'] = dcoingecko_formatado['Data'].str[11:]
dcoingecko_formatado['Data'] = dcoingecko_formatado['Data'].str[:10]
```

```python
dcoingecko_formatado['mês'] = dcoingecko_formatado['Data'].dt.month
dcoingecko_formatado['ano'] = dcoingecko_formatado['Data'].dt.year
```
